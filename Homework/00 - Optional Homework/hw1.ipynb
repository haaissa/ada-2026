```python
import csv
import pandas as pd

with open("interactive_data.csv", "r") as f:
    data = csv.reader(f)
    header = next(data)
    df = pd.DataFrame(data, columns=header)
print(df.head())
```

```python
def print_proportion(a, b, intent):
    proportion = a/b
    print("proportion of %s %f" % (intent, proportion))
```

```python
# nb of gun deaths that are suicides
def intent_proportion(df, intent):
    df_intent = df[df["Intent"] != "None selected"]
    df_intent_suicide = df_intent[df_intent["Intent"] == intent]
    
    total_intent_sum = df_intent["Deaths"].astype(int).sum()
    total_intent_suicide_sum = df_intent_suicide["Deaths"].astype(int).sum()
    
    print_proportion(total_intent_suicide_sum,total_intent_sum, intent)
    return df_intent_suicide

df_suicide= intent_proportion(df, "Suicide")
```

```python
# More than 85 percent of suicide victims are male.
df_suicide_filtered = df_suicide[df_suicide["Gender"] != "None selected"]
df_suicide_male = df_suicide_filtered[df_suicide_filtered["Gender"] == "Male"]

total_suicide_sum = df_suicide_filtered["Deaths"].astype(int).sum()
total_suicide_male_sum = df_suicide_male["Deaths"].astype(int).sum()

print_proportion(total_suicide_male_sum, total_suicide_sum, "Suicide")
```

```python
# Around a third of all gun deaths are homicides.
df_homicide = intent_proportion(df, "Homicide")
```

```python
# Around two-thirds of homicide victims who are males in the age-group of 15--34 are black.
df_homicide_filtered = df_homicide[(df_homicide["Age"] != "None selected") & (df_homicide["Race"] != "None selected") & (df_homicide["Gender"] != "None selected")]
df_middleAged_male = df_homicide_filtered[(df_homicide_filtered["Age"] == "15 - 34") & (df_homicide_filtered["Gender"] == "Male")]
df_black = df_middleAged_male[df_middleAged_male["Race"] == "Black"]

total = df_middleAged_male["Deaths"].astype(int).sum()
total_black = df_black["Deaths"].astype(int).sum()

print_proportion(total_black, total, "Black male in the age-group of 15--34 Homicide")
```

```python
# Women constitue only 15 percent of the total homicide victims.
df_homicide_filtered = df_homicide[df_homicide["Gender"] != "None selected"]

total_homicide_women = df_homicide[df_homicide["Gender"] == "Female"]["Deaths"].astype(int).sum()
total_homicide = df_homicide_filtered["Deaths"].astype(int).sum()

print_proportion(total_homicide_women, total_homicide, "Women Homicide")
```
